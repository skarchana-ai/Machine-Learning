Feature Engineering

Step 1: Missing-Value Handling

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
data={
    "Name":["Akhil","Rahul","Anu","Meera","John"],
    "Age":[20,None,22,21,None],
    "Marks":[85,90,None,75,80],
    "City":["Kochi","Calicut",None,"Kochi","Calicut"]

}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

Check missing values in each column

```python
df.isnull().sum()
```

Convert any python "None"or string"None"into actual mathematical NaN values

```python
df["City"]=df["City"].replace(["None",None],np.nan)
```

```python
df
```

```python
num_imputer=SimpleImputer(strategy='median')
df[["Age","Marks"]]=num_imputer.fit_transform(df[["Age","Marks"]])

cat_imputer=SimpleImputer(strategy="most_frequent")
df[["City"]]=cat_imputer.fit_transform(df[["City"]])
```

```python
df
```

Min-Max Scaling(Normalization)

```python
data = {
    'Name': ['Akhil', 'Rahul', 'John', 'Anu', 'David'],
    'Age': [20, 22, 21, 23, 24],
    'Study_Hours': [2, 4, 3, 5, 6],
    'Marks': [450, 780, 650, 900, 850],
    'Salary': [15000, 25000, 20000, 30000, 28000]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
from sklearn.preprocessing import MinMaxScaler

#Create MinMaxScaler object
scaler=MinMaxScaler()

#select numerical columns
columns=['Age','Study_Hours','Marks','Salary']

#Apply Min-Max Scaling
scaled_data=scaler.fit_transform(df[columns])
```

```python
scaled_data
```

```python
scaled_df=pd.DataFrame(
    scaled_data,
    columns=columns
)
```

```python
scaled_df
```

Step 1: Missing-Value Handling

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

Check missing values in each column

Step 4 - Creating a Feature

 Feature Creation is an important techinique in Feature Engineering where we create new columns(features) from existing columns in a dataset to help Machine Learning models learn useful patterns.

 In simple words:
 We alraedy have some information in our dataset. Instead of using only the existing columns, we can combine or transform them to create new meaningful information.

 Sometimes, combining two columns into one provides valuable new information.

```python
data = {
    'Employee': ['Akhil', 'Rahul', 'John', 'Anu', 'David'],

    'Age': [25, 30, 28, 35, 40],

    'Experience': [2, 5, 4, 10, 15],

    'Monthly_Salary': [25000, 40000, 35000, 60000, 80000],

    'Working_Hours': [8, 9, 7, 8, 10],

    'Projects': [3, 5, 4, 8, 12]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

Create Annual Salary

```python
df['Annual_Salary']=df['Monthly_Salary']*17
```

```python
df
```

Create Weekly Working Hours

```python
df['Weekly_Hours']=df['Working_Hours']*5
```

```python
df
```

Create project per year

```python
df['Projects_Per_Year']=(
    df['Projects']/df['Experience']
)
```

```python
df
```

Create annual working hours

```python
df['Annual_Working_Hours']=df['Weekly_Hours']*52
```

```python
df
```

Create Experience Level

```python
df['Experience_Level']=np.where(
    df['Experience']<3,
    'Beginner',
    np.where(
        df['Experience']<=7,
        'Intermediate',
        'Experienced'
    )
)
```

Outlier Detection and Treatment

Outliers are values that are unusually high or low compared with the other values in the dataset.

```python
data = {
    'Name': ['Akhil', 'Rahul', 'John', 'Anu', 'David',
             'Sara', 'Peter', 'Meera', 'Arun', 'Vishnu'],
    'Age': [22, 25, 28, 30, 27, 29, 26, 31, 150, 24],
    'Salary': [25000, 30000, 35000, 40000, 32000,
               38000, 28000, 42000, 39000, 500000]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

Detect Outliers using the IQR method.

One common method for detecting outliers is the Interquartile Range (IQR) method.

First,Understand these three values:

*   Q1(First Quartile)
*   Q2(Second Quartile)
*   Q3(Third Quartile)






```python
#Calculate Q1 and Q3
Q1=df['Age'].quantile(0.25)
Q3=df['Age'].quantile(0.75)

#Calculate IQR
IQR=Q3-Q1

#Calculate boundaries
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)

print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

Detect Outliers

```python
print(df.head())
```

```python
outliers=df[(df['Age']<lower) | (df['Age']>upper)]
```

```python
outliers
```

Method 3 :Capping (Winsorisation)

Capping means replacing outliers with the nearest acceptable boundary instaed of removing them.

For example:

*   If a value is above the upper boundary, replace it with the upper boundary.
*   If a value is below the lower boundary, replace it with the lower boundary.



```python
#Calculate IQR boundaries
Q1=df['Age'].quantile(0.25)
Q3=df['Age'].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3=1.5*IQR

#Apply capping
df['Age']=df['Age'].clip(
    lower=lower,
    upper=upper
)

print(df)
```

Method 4: Log Transformation

Log transformation reduces the impact of very large numerical values. It is useful for highly right-skewed data

```python
import numpy as np

#Apply log transformation
df['Salary_Log']=np.log1p(df['Salary'])

print(df[['Salary', 'Salary_Log']])
```

Data and Time Feature Engineering

Date and Time Feature Engineering is the process of extracting useful information from date and time columns to create new features that can help machine learning models identify patterns.

```python
data = {
    'Order_ID': [101, 102, 103, 104, 105, 106],
    'Customer': ['Akhil', 'Rahul', 'John', 'Anu', 'David', 'Sara'],
    'Order_Date': [
        '2026-01-15 09:30:00',
        '2026-02-20 14:45:00',
        '2026-03-10 18:20:00',
        '2026-04-05 11:15:00',
        '2026-05-25 20:30:00',
        '2026-06-12 08:00:00'
    ],
    'Amount': [2500, 3500, 1800, 4200, 5000, 2800]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

Convert the column into DateTime format

Before extracting date and time information, we need to convert the column into Pandas DateTime format.

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'])
```

```python
print(df.dtypes)
```

Extract Year, Month and Day

```python
#Extract year
df['Year']=df['Order_Date'].dt.year

#Extract Month
df['Month']=df['Order_Date'].dt.month

#Extract day
df['Day']=df ['Order_Date'].dt.day

print(df[['Order_Date','Year','Month','Day']])
```

Extract Day of the week

```python
df['Day_Name']=df['Order_Date'].dt.day_name()
```

```python
print(df[['Order_Date','Day_Name']])
```

Extract Hour,Minute and Second

```python
#Extract hour
df['Hour']=df['Order_Date'].dt.hour

#Extract minute
df['Minute']=df['Order_Date'].dt.minute

#Extract Second
df['Second']=df['Order_Date'].dt.second

print(df[['Order_Date','Hour','Minute','Second']])
```

We can also extract time-realted information.

```python
#Extract hour
df['Hour']=df['Order_Date'].dt.hour

#Extract minute
df['Minute']=df['Order_Date'].dt.minute

#Extract second
df['Second'] = df['Order_Date'].dt.second

print(df[['Order_Date','Hour','Minute','Second']])
```
